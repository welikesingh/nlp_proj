```python



# Step 1: Import Required Libraries
# ===============================================================

import pandas as pd
import joblib

# Used for splitting the dataset

from sklearn.model_selection import train_test_split

# Used for converting text into numerical vectors

from sklearn.feature_extraction.text import TfidfVectorizer

# Machine Learning Algorithm

from sklearn.naive_bayes import MultinomialNB

# Model Evaluation

from sklearn.metrics import accuracy_score, classification_report

# ===============================================================
# Step 2: Create Dataset
# ===============================================================

data = {

    "News":[

        "Government launches new healthcare scheme for poor families",

        "Scientists discover a new planet similar to Earth",

        "India wins the cricket world cup after a thrilling match",

        "Stock market reaches all time high this week",

        "New education policy approved by parliament",

        "Aliens landed in New York yesterday",

        "Drinking hot water cures every disease instantly",

        "The moon is made entirely of gold",

        "Dinosaurs are still living in the Amazon forest",

        "Humans can survive without oxygen for ten hours"

    ],

    "Label":[

        "Real",

        "Real",

        "Real",

        "Real",

        "Real",

        "Fake",

        "Fake",

        "Fake",

        "Fake",

        "Fake"

    ]

}

# ===============================================================
# Step 3: Convert Dictionary into DataFrame
# ===============================================================

df = pd.DataFrame(data)

print("======================================")
print("Dataset")
print("======================================")

print(df)

# ===============================================================
# Step 4: Separate Features and Target
# ===============================================================

X = df["News"]

# Input Text

y = df["Label"]

# Output Label

# ===============================================================
# Step 5: Convert Text into Numbers
# ===============================================================

vectorizer = TfidfVectorizer(

    stop_words="english",

    lowercase=True

)

# Learn Vocabulary

X_vector = vectorizer.fit_transform(X)

print("\nTotal Features :", len(vectorizer.get_feature_names_out()))

# ===============================================================
# Step 6: Split Dataset
# ===============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X_vector,

    y,

    test_size=0.20,

    random_state=42

)

print("\nTraining Samples :", X_train.shape[0])

print("Testing Samples :", X_test.shape[0])

# ===============================================================
# Step 7: Create Machine Learning Model
# ===============================================================

model = MultinomialNB()

# ===============================================================
# Step 8: Train Model
# ===============================================================

model.fit(

    X_train,

    y_train

)

print("\nModel Training Completed Successfully")

# ===============================================================
# Step 9: Prediction on Test Data
# ===============================================================

prediction = model.predict(

    X_test

)

# ===============================================================
# Step 10: Evaluate Model
# ===============================================================

accuracy = accuracy_score(

    y_test,

    prediction

)

print("\n===================================")

print("Model Accuracy")

print("===================================")

print(accuracy)

print("\nClassification Report")

print(classification_report(

    y_test,

    prediction

))

# ===============================================================
# Step 11: Test Custom News
# ===============================================================

news = [

    input("\nEnter News : ")

]

news_vector = vectorizer.transform(

    news

)

result = model.predict(

    news_vector

)

print("\nPrediction :", result[0])

# ===============================================================
# Step 12: Save Model
# ===============================================================

joblib.dump(

    model,

    "model.pkl"

)

joblib.dump(

    vectorizer,

    "vectorizer.pkl"

)

print("\nModel Saved Successfully")

print("model.pkl")

print("vectorizer.pkl")

# ===============================================================
# Step 13: Show Vocabulary Size
# ===============================================================

print("\nVocabulary Size :")

print(len(vectorizer.vocabulary_))

# ===============================================================
# Step 14: Display Some Important Words
# ===============================================================

print("\nSample Words :")

print(vectorizer.get_feature_names_out()[:20])

# ===============================================================
# Step 15: Project Completed
# ===============================================================

print("\n======================================")

print("Fake News Detection Model is Ready")

print("======================================")
```
