```python

# Step 1: Import Required Libraries
# ===============================================================

import pandas as pd
import joblib

from sklearn.model_selection import train_test_split

from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import accuracy_score, classification_report

# ===============================================================
# Step 2: Create Dataset
# ===============================================================

data = {

    "News":[

        # Sports

        "India won the cricket world cup",
        "Virat Kohli scored a century",
        "Football team won the championship",
        "Olympics will begin next month",

        # Politics

        "Government introduced a new tax policy",
        "Prime Minister addressed the nation",
        "Parliament passed the education bill",
        "Election results announced today",

        # Technology

        "Artificial Intelligence is transforming industries",
        "Google launches new AI model",
        "Python is popular for data science",
        "Cyber security threats are increasing",

        # Business

        "Stock market reached a record high",
        "Company reported huge quarterly profit",
        "Gold prices increased today",
        "New startup raised funding",

        # Entertainment

        "The new movie broke box office records",
        "Actor won the best performance award",
        "Music concert attracted thousands",
        "New web series released today"

    ],

    "Category":[

        "Sports",
        "Sports",
        "Sports",
        "Sports",

        "Politics",
        "Politics",
        "Politics",
        "Politics",

        "Technology",
        "Technology",
        "Technology",
        "Technology",

        "Business",
        "Business",
        "Business",
        "Business",

        "Entertainment",
        "Entertainment",
        "Entertainment",
        "Entertainment"

    ]

}

# ===============================================================
# Step 3: Create DataFrame
# ===============================================================

df = pd.DataFrame(data)

print("===================================")

print("News Category Dataset")

print("===================================")

print(df)

# ===============================================================
# Step 4: Separate Features and Target
# ===============================================================

X = df["News"]

y = df["Category"]

# ===============================================================
# Step 5: TF-IDF Vectorization
# ===============================================================

vectorizer = TfidfVectorizer(

    stop_words="english",

    lowercase=True

)

X_vector = vectorizer.fit_transform(X)

print("\nVocabulary Size :", len(vectorizer.get_feature_names_out()))

# ===============================================================
# Step 6: Train-Test Split
# ===============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X_vector,

    y,

    test_size=0.20,

    random_state=42

)

# ===============================================================
# Step 7: Create Model
# ===============================================================

model = MultinomialNB()

# ===============================================================
# Step 8: Train Model
# ===============================================================

model.fit(

    X_train,

    y_train

)

print("\nModel Trained Successfully")

# ===============================================================
# Step 9: Prediction
# ===============================================================

prediction = model.predict(

    X_test

)

# ===============================================================
# Step 10: Evaluate Model
# ===============================================================

accuracy = accuracy_score(

    y_test,

    prediction

)

print("\nAccuracy :", accuracy)

print("\nClassification Report")

print(classification_report(

    y_test,

    prediction

))

# ===============================================================
# Step 11: User Prediction
# ===============================================================

news = [

    input("\nEnter News Headline : ")

]

news_vector = vectorizer.transform(

    news

)

result = model.predict(

    news_vector

)

print("\nPredicted Category :", result[0])

# ===============================================================
# Step 12: Save Model
# ===============================================================

joblib.dump(

    model,

    "pmodel.pkl"

)

joblib.dump(

    vectorizer,

    "pvectorizer.pkl"

)

print("\nModel Saved Successfully")

# ===============================================================
# Step 13: Vocabulary Size
# ===============================================================

print("\nVocabulary Size :")

print(len(vectorizer.vocabulary_))

# ===============================================================
# Step 14: Display Sample Features
# ===============================================================

print("\nSample Features :")

print(vectorizer.get_feature_names_out()[:30])

# ===============================================================
# Step 15: Project Completed
# ===============================================================

print("\n===================================")

print("News Category Classification Model is Ready")

print("===================================")
```
